## Milestone 1 – cProfile Analysis

### 1. Which function takes most total time?

From the naive profile output:

- `mandelbrot_naive` total time: **5.034 seconds**
- `mandelbrot_point` cumulative time: **4.310 seconds**

The function that takes most total time is **`mandelbrot_point`**.  
It accounts for the majority of execution time in the naive implementation.  
This indicates that the inner iteration loop of the Mandelbrot computation is the main performance bottleneck.

---

### 2. Are there functions called surprisingly many times?

Yes.

The function `mandelbrot_point` is called **262,144 times**, which corresponds exactly to:

512 × 512 = 262,144 pixels.

This means the function is executed once per pixel in the grid.  
The extremely large number of Python-level function calls explains why the naive implementation is slow.

---

### 3. How does the NumPy profile compare to naive?

The NumPy profile shows:

- Only **74 function calls**
- Total runtime of **0.619 seconds**

Compared to the naive version:

- Naive: **262,175 function calls**
- NumPy: **74 function calls**

This dramatic reduction in Python-level calls shows that most work in the NumPy implementation happens inside optimized compiled C routines (e.g., broadcasting and vectorized operations).

Therefore, NumPy is significantly faster because it avoids explicit Python loops and reduces interpreter overhead.

```python
print(r"""
File: week3/mandelbrot_week3_m2.py
Function: mandelbrot_naive at line 21

Total time: 9.41077 s

Line #      Hits        Time  Per Hit   % Time  Line Contents
==============================================================
26             1       174.6    174.6      0.0   x = np.linspace(xmin, xmax, width)
27             1        55.7     55.7      0.0   y = np.linspace(ymin, ymax, height)
29             1        28.4     28.4      0.0   result = np.zeros((height, width), dtype=np.int32)
31           513       150.0      0.3      0.0   for i in range(height):
32        262656     76134.2      0.3      0.8       for j in range(width):
33        262144    687667.1      2.6      7.3           c = x[j] + 1j * y[i]
34        262144   8646559.2     33.0     91.9           result[i, j] = mandelbrot_point(c, max_iter)
36             1         1.1      1.1      0.0   return result
""")
```

## Milestone 2 – Deep Profiling

### 1. cProfile: Naive vs NumPy

Naive implementation:
- 262,175 function calls
- 5.034 s total runtime
- 262,144 calls to `mandelbrot_point`

NumPy implementation:
- 74 function calls
- 0.619 s total runtime

The naive version spends most time in Python-level loops and repeated function calls.  
The NumPy version shifts work to compiled routines, reducing interpreter overhead and runtime.

---

### 2. line_profiler (naive)

Dominant lines:

- `result[i, j] = mandelbrot_point(c, max_iter)` → **91.9%**
- `c = x[j] + 1j * y[i]` → **7.3%**

Inner loop total ≈ **99.2%** of runtime.

The bottleneck is clearly the repeated call to `mandelbrot_point` inside the nested loops.

---

### 3. Why is NumPy faster?

NumPy avoids explicit Python loops and repeated function calls.  
Most computation runs in optimized compiled code, resulting in far fewer Python-level calls (74 vs 262,175).

---

### 4. How to make naive faster?

- Remove Python-level inner loop overhead
- Avoid per-pixel Python function calls
- Use NumPy vectorization or Numba JIT

```python
print(r"""
Naive: 4.785 s
NumPy: 0.575 s
Numba (serial): 0.019 s

Speedups vs Naive:
NumPy speedup: 8.32x
Numba speedup: 251.84x
""")
```

## Milestone 3 – Numba JIT (serial)

Numba significantly improves performance compared to naive Python.

- Naive: 4.785 s  
- NumPy: 0.575 s  
- Numba: 0.019 s  

Numba achieves ~252× speedup over naive Python.

Numba is faster because it compiles the nested loops and the inner `while` loop to machine code, eliminating Python interpreter overhead. The Mandelbrot algorithm benefits from early exit in the loop, which Numba handles efficiently in compiled form.

## Milestone 4 – Data type precision vs performance

### Speed (float32 vs float64)
On my CPU, float32 is **not faster** than float64 in this test:
- float32: **0.074575 s**
- float64: **0.074407 s**
Difference: **0.000168 s** (~**0.23%**), essentially the same (within noise).

### float16 (NumPy / Numba note)
- **NumPy float16:** In general, NumPy often promotes float16 operations to float32 internally, so float16 may **not** be faster than float32 on CPU (and can be similar or slower).
- **Numba float16:** On my setup, Numba **refused to compile native float16** for this kernel. I therefore used **float16 quantization** (round inputs to float16, compute in float32) to evaluate float16 precision effects. Runtime was **0.074424 s**, essentially the same as float32.

### Visual quality (zoomed comparison)
In the chosen zoom region, float32 and float64 are visually indistinguishable. Float16 (quantized) also appears very similar at this scale; artefacts would be more likely at deeper zoom / higher max_iter.

### Recommendation
For CPU production use, I would choose **float32**: it provides sufficient precision for the image at this zoom level and shows **no meaningful speed disadvantage** compared with float64 in my tests. I would avoid float16 on CPU because it provides **no speed benefit here** and reduces numerical precision; precision choices are expected to matter more on GPUs.

## Milestone 5 – Parallel Numba + Scaling

I parallelized the outer loop over rows using `@njit(parallel=True)` and `prange`.

**Result:** Parallel execution reduced runtime from **0.832 s** to **0.304 s**, giving **2.73×** speedup.

**Scaling:** Speedup increases slightly with problem size:
- 512²: **2.63×**
- 1024²: **2.85×**
- 2048²: **2.88×**

Speedup is sub-linear and saturates around ~2.9× due to parallel overhead, memory bandwidth limits, and load imbalance (different pixels require different iteration counts).
